## Sumowanie ciągu geometrycznego

Suma $n$ pierwszych wyrazów ciągu geometrycznego o pierwszym wyrazie $a_1$ oraz ilorazie $q \neq 1$ wyraża się wzorem:

$$S_n = a_1 \cdot \frac{1 - q^n}{1 - q}$$

Przykład w języku Python:

```python
def suma_ciagu_geom(a1, q, n):
    if q == 1:
        return a1 * n
    return a1 * (1 - q**n) / (1 - q)

# Przykład dla a1=2, q=3, n=5
print("Suma ciągu geometrycznego:", suma_ciagu_geom(2, 3, 5))
```

```python
# Wykres y = k^2 dla k = 1, ..., 20 oraz zapis do wykres.png
import matplotlib.pyplot as plt

k = list(range(1, 21))
y = [i**2 for i in k]

plt.figure(figsize=(8, 5))
plt.plot(k, y, 'o-', color='blue', label='$y = k^2$')
plt.title("Wykres $y = k^2$ dla $k \in [1, 20]$")
plt.xlabel("k")
plt.ylabel("y")
plt.grid(True)
plt.legend()

# Zapis wykresu
plt.savefig("wykres.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Animacja wahadła, zapis do video.mp4 i osadzenie w notebooku
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import Video

fig, ax = plt.subplots(figsize=(5, 5))
ax.set_xlim(-1.2, 1.2)
ax.set_ylim(-1.2, 0.2)
ax.set_aspect('equal')
ax.grid(True)

line, = ax.plot([], [], 'o-', lw=2, color='red', markersize=10)

def init():
    line.set_data([], [])
    return line,

def animate(i):
    t = i * 0.1
    theta = 0.5 * np.cos(2 * np.pi * t / 2)
    x = np.sin(theta)
    y = -np.cos(theta)
    line.set_data([0, x], [0, y])
    return line,

anim = FuncAnimation(fig, animate, init_func=init, frames=60, interval=50, blit=True)

# Zapis do pliku video.mp4
anim.save('video.mp4', writer='ffmpeg', fps=20)
plt.close()

# Osadzenie wideo w notebooku
Video("video.mp4", embed=True)
```

```python
# Zmodyfikowana krzywa parametryczna 3D i zapis do parametric_curve.png
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Zmodyfikowane równania parametryczne (helisa / rurka parametryczna)
t = np.linspace(0, 4 * np.pi, 500)
x = (1 + np.cos(t)) * np.cos(t)
y = (1 + np.cos(t)) * np.sin(t)
z = np.sin(t)

# Wykres 3D
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')
ax.plot(x, y, z, label="Krzywa parametryczna 3D", color="purple", lw=2)
ax.set_title("3D Parametric Curve", fontsize=14)
ax.set_xlabel("x(t)")
ax.set_ylabel("y(t)")
ax.set_zlabel("z(t)")
ax.legend()

# Zapis do parametric_curve.png
plt.savefig("parametric_curve.png", dpi=300, bbox_inches='tight')
plt.show()
```
